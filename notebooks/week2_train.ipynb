```python
import torch, json, numpy as np, pandas as pd, matplotlib.pyplot as plt

def load(path):
    sd = torch.load(path, map_location="cpu")
    return sd.get("state_dict", sd)

names = {"fmnist": "Fashion-MNIST", "cifar10": "CIFAR-10"}
nets = {d: load(f"models/vgg_{d}_fp32.pt") for d in names}

# a) weights per layer (conv = 4-D tensors, linear = 2-D)
fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharex=True)
for ax, (d, sd) in zip(axes, nets.items()):
    s = {k.replace(".weight", ""): v.numel() for k, v in sd.items()
         if k.endswith("weight") and v.ndim in (2, 4)}
    pd.Series(s).plot.barh(ax=ax)
    ax.invert_yaxis(); ax.set_title(f"VGG on {names[d]}"); ax.set_xlabel("weights")
plt.tight_layout(); plt.savefig("figures/vgg_weights_per_layer.png", dpi=150); plt.show()

# b) weight sizes: how many are >= 1, which decides what bit 14 can do
fig, ax = plt.subplots(figsize=(8, 4))
for d, sd in nets.items():
    w = torch.cat([v.flatten() for v in sd.values() if v.ndim in (2, 4)]).half().float().abs()
    w = w[w > 0].numpy()
    print(f"{names[d]}: max |w| = {w.max():.3f}, share with |w| >= 1 = {(w >= 1).mean():.5f}")
    ax.hist(np.log2(w), bins=80, alpha=0.6, density=True, label=names[d])
ax.axvline(0, color="k", ls="--"); ax.set_xlabel("log2 |w|"); ax.set_ylabel("density"); ax.legend()
plt.tight_layout(); plt.savefig("figures/vgg_weight_magnitudes.png", dpi=150); plt.show()

# c) training results table, straight from the logs
rows = []
for d in names:
    m = json.load(open(f"logs/train_vgg_{d}.json"))
    rows.append(dict(dataset=names[d], params=m["parameters"], acc_fp32=m["acc_fp32"],
                     acc_fp16=m["acc_fp16"], sha=m["weights_sha256"][:12]))
print(pd.DataFrame(rows).to_string(index=False))
```
